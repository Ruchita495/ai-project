```python
import os
from dotenv import load_dotenv
def validate_environment_setup() -> None:
    """
    Verifies that all required API keys are properly configured in the environment.
    """
    load_dotenv()
    key_config = {
        "OPENAI_API_KEY": "sk-"
    }

    print("--- Environment Configuration Check ---")
    for key, value in key_config.items():
        if not value:
            print(f"{key:<20}: Not found in .env.")
            continue

    if not value.startswith(value):
        print(f"{key:<20}: Found, but expected prefix '{value}'")
    elif not value.strip() != value:
        print(f"{key:<20}: Found, but conatins hidden whitespace")
    else:
        print(f"{key:<20}: Validated (end with ...{value[-4:]})")
validate_environment_setup()
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

def quick_test_frontire_models():
    #1. Test OpenAI
    load_dotenv()
    print("Testing OpenAI...")
    openai_client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
    response = openai_client.chat.completions.create(
        model="gpt-5-mini",
        messages=[
                {"role": "user", "content": "What is the capital of the moon?"}
            ],
        )
    print(response)
    print(response.choices[0].message.content)
   

quick_test_frontire_models()
```
